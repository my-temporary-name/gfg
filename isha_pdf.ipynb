```python
import fitz  # PyMuPDF

def extract_hindi_text_from_pdf(file_path):
    # Open the PDF file
    doc = fitz.open(file_path)
    hindi_text = ""
    
    # Iterate through each page of the PDF
    for page in doc:
        # Extract text from the current page
        text = page.get_text()
        
        # Append only Hindi characters; Hindi Unicode range is U+0900 to U+097F
        hindi_text += "".join([char for char in text if '\u0900' <= char <= '\u097F'])
    
    return hindi_text

# Path to the PDF file
file_path = 'A:\coding\gfg\dummy1.pdf'  # Replace this with your actual file path
hindi_text = extract_hindi_text_from_pdf(file_path)
print(hindi_text)
```

```python
import fitz  # Import PyMuPDF

def convert_pdf_to_images_fitz(pdf_path):
    doc = fitz.open(pdf_path)  # Open the PDF file
    for page_number in range(len(doc)):  # Iterate through each page
        page = doc.load_page(page_number)  # Load the current page
        pix = page.get_pixmap()  # Render page to an image
        pix.save(f"page_{page_number}.png")  # Save the image as PNG

# Path to the PDF file
pdf_path = 'A:\coding\gfg\isha12.pdf'  # Replace with your PDF file path
convert_pdf_to_images_fitz(pdf_path)
```
